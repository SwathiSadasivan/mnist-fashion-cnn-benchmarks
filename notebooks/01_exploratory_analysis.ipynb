```python
import torch
import matplotlib.pyplot as plt
import numpy as np
from src.dataset import get_dataloaders

# Load Fashion-MNIST training data
train_loader, _ = get_dataloaders(dataset_name="fashion_mnist", batch_size=64)

# Get a batch of images and labels
images, labels = next(iter(train_loader))

# Class names mapping
class_names = [
    "T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]

# Plot sample grid
fig, axes = plt.subplots(2, 5, figsize=(12, 6))
for i, ax in enumerate(axes.flat):
    img = images[i].squeeze().numpy()
    label = labels[i].item()
    ax.imshow(img, cmap="gray")
    ax.set_title(class_names[label])
    ax.axis("off")

plt.tight_layout()
plt.show()

# Print tensor specs
print(f"Batch Tensor Shape: {images.shape}")
print(f"Single Image Shape: {images[0].shape}")
print(f"Pixel Value Range: [{images.min():.2f}, {images.max():.2f}]")
```
